```python
import os
import sys
import subprocess
import random

subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "ultralytics", "opencv-python-headless", "matplotlib"])

import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

random.seed(42)
rng = np.random.default_rng(42)

KLASY_LOGO = ["Mastercard", "Target", "YouTube", "Adidas"]
KATALOG_BAZOWY = "zestaw_danych_logo"


def przygotuj_katalogi():
    podkatalogi = [
        "images/train",
        "images/val",
        "images/test_surowe",
        "images/test_przetworzone",
        "labels/train",
        "labels/val",
    ]
    for pod in podkatalogi:
        os.makedirs(os.path.join(KATALOG_BAZOWY, pod), exist_ok=True)


def rysuj_zaokraglony_prostokat(obraz, pt1, pt2, promien, kolor):
    x1, y1 = pt1
    x2, y2 = pt2
    cv2.rectangle(obraz, (x1 + promien, y1), (x2 - promien, y2), kolor, -1)
    cv2.rectangle(obraz, (x1, y1 + promien), (x2, y2 - promien), kolor, -1)
    cv2.circle(obraz, (x1 + promien, y1 + promien), promien, kolor, -1)
    cv2.circle(obraz, (x2 - promien, y1 + promien), promien, kolor, -1)
    cv2.circle(obraz, (x1 + promien, y2 - promien), promien, kolor, -1)
    cv2.circle(obraz, (x2 - promien, y2 - promien), promien, kolor, -1)


def generuj_szablon_logo(id_klasy):
    rozmiar = 160
    kanwa = np.zeros((rozmiar, rozmiar, 4), dtype=np.uint8)

    if id_klasy == 0:
        lewy = np.zeros((rozmiar, rozmiar), dtype=np.uint8)
        prawy = np.zeros((rozmiar, rozmiar), dtype=np.uint8)
        cv2.circle(lewy, (58, 80), 44, 255, -1, lineType=cv2.LINE_AA)
        cv2.circle(prawy, (102, 80), 44, 255, -1, lineType=cv2.LINE_AA)
        kanwa[lewy > 0] = (27, 0, 235, 255)
        kanwa[prawy > 0] = (27, 158, 247, 255)
        wspolny = (lewy > 0) & (prawy > 0)
        kanwa[wspolny] = (27, 92, 244, 255)

    elif id_klasy == 1:
        cv2.circle(kanwa, (80, 80), 56, (0, 0, 204, 255), -1, lineType=cv2.LINE_AA)
        cv2.circle(kanwa, (80, 80), 38, (255, 255, 255, 255), -1, lineType=cv2.LINE_AA)
        cv2.circle(kanwa, (80, 80), 19, (0, 0, 204, 255), -1, lineType=cv2.LINE_AA)

    elif id_klasy == 2:
        rysuj_zaokraglony_prostokat(kanwa, (22, 42), (138, 118), 18, (0, 0, 245, 255))
        trojkat = np.array([[68, 60], [68, 100], [104, 80]], dtype=np.int32)
        cv2.fillPoly(kanwa, [trojkat], (255, 255, 255, 255), lineType=cv2.LINE_AA)

    elif id_klasy == 3:
        rysuj_zaokraglony_prostokat(kanwa, (20, 24), (140, 136), 14, (245, 245, 245, 240))
        pasek_1 = np.array([[42, 118], [60, 118], [50, 96], [32, 96]], dtype=np.int32)
        pasek_2 = np.array([[70, 118], [88, 118], [70, 72], [52, 72]], dtype=np.int32)
        pasek_3 = np.array([[98, 118], [116, 118], [90, 48], [72, 48]], dtype=np.int32)
        cv2.fillPoly(kanwa, [pasek_1, pasek_2, pasek_3], (20, 20, 20, 255), lineType=cv2.LINE_AA)

    return kanwa


def generuj_tlo(rozmiar=416):
    baza = rng.integers(50, 190, size=(16, 16, 3), dtype=np.uint8)
    tlo = cv2.resize(baza, (rozmiar, rozmiar), interpolation=cv2.INTER_CUBIC)
    for _ in range(12):
        p1 = (int(rng.integers(0, rozmiar)), int(rng.integers(0, rozmiar)))
        p2 = (int(rng.integers(0, rozmiar)), int(rng.integers(0, rozmiar)))
        kolor = tuple(int(c) for c in rng.integers(40, 210, size=3))
        cv2.line(tlo, p1, p2, kolor, int(rng.integers(1, 4)), lineType=cv2.LINE_AA)
    return cv2.GaussianBlur(tlo, (7, 7), 1.5)


def degraduj_obraz(obraz_bgr):
    kontrast = float(rng.uniform(0.50, 0.72))
    jasnosc = float(rng.uniform(35.0, 62.0))
    wyblakly = np.clip(obraz_bgr.astype(np.float32) * kontrast + jasnosc, 0, 255).astype(np.uint8)
    rozmyty = cv2.GaussianBlur(wyblakly, (3, 3), 0.9)
    szum = rng.normal(0.0, 10.0, rozmyty.shape).astype(np.float32)
    return np.clip(rozmyty.astype(np.float32) + szum, 0, 255).astype(np.uint8)


def przetwarzanie_wstepne(obraz_bgr):
    odszumiony = cv2.bilateralFilter(obraz_bgr, d=5, sigmaColor=45, sigmaSpace=45)
    lab = cv2.cvtColor(odszumiony, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)
    clahe = cv2.createCLAHE(clipLimit=2.8, tileGridSize=(8, 8))
    l_clahe = clahe.apply(l)
    poprawiony_kontrast = cv2.cvtColor(cv2.merge((l_clahe, a, b)), cv2.COLOR_LAB2BGR)
    rozmyty = cv2.GaussianBlur(poprawiony_kontrast, (0, 0), sigmaX=1.4)
    wyostrzony = cv2.addWeighted(poprawiony_kontrast, 1.45, rozmyty, -0.45, 0)
    return cv2.normalize(wyostrzony, None, 0, 255, cv2.NORM_MINMAX)


def sprawdz_kolizje(nowy_box, istniejace_boxy):
    nx1, ny1, nx2, ny2 = nowy_box
    for bx1, by1, bx2, by2 in istniejace_boxy:
        if nx1 < bx2 + 8 and nx2 + 8 > bx1 and ny1 < by2 + 8 and ny2 + 8 > by1:
            return True
    return False


def generuj_probke(rozmiar=416):
    tlo = generuj_tlo(rozmiar)
    liczba_obiektow = int(rng.integers(1, 4))
    zajete_obszary = []
    etykiety_yolo = []

    for _ in range(liczba_obiektow):
        id_klasy = int(rng.integers(0, len(KLASY_LOGO)))
        szablon = generuj_szablon_logo(id_klasy)
        kat = float(rng.uniform(-22.0, 22.0))
        skala = float(rng.uniform(0.38, 0.72))
        macierz = cv2.getRotationMatrix2D((80, 80), kat, skala)
        obrocony = cv2.warpAffine(
            szablon,
            macierz,
            (160, 160),
            flags=cv2.INTER_LINEAR,
            borderMode=cv2.BORDER_CONSTANT,
            borderValue=(0, 0, 0, 0),
        )
        ys, xs = np.where(obrocony[:, :, 3] > 25)
        if len(xs) == 0 or len(ys) == 0:
            continue
        x1, x2 = int(xs.min()), int(xs.max())
        y1, y2 = int(ys.min()), int(ys.max())
        wycinek = obrocony[y1 : y2 + 1, x1 : x2 + 1]
        wys, szer = wycinek.shape[:2]

        for _ in range(30):
            px = int(rng.integers(10, rozmiar - szer - 10))
            py = int(rng.integers(10, rozmiar - wys - 10))
            kandydat = (px, py, px + szer, py + wys)
            if not sprawdz_kolizje(kandydat, zajete_obszary):
                zajete_obszary.append(kandydat)
                alfa = (wycinek[:, :, 3:4].astype(np.float32)) / 255.0
                fg = wycinek[:, :, :3].astype(np.float32)
                bg = tlo[py : py + wys, px : px + szer].astype(np.float32)
                tlo[py : py + wys, px : px + szer] = np.clip(fg * alfa + bg * (1.0 - alfa), 0, 255).astype(np.uint8)

                xc = (px + szer / 2.0) / rozmiar
                yc = (py + wys / 2.0) / rozmiar
                wn = szer / rozmiar
                hn = wys / rozmiar
                etykiety_yolo.append(f"{id_klasy} {xc:.6f} {yc:.6f} {wn:.6f} {hn:.6f}")
                break

    surowy = degraduj_obraz(tlo)
    przetworzony = przetwarzanie_wstepne(surowy)
    return surowy, przetworzony, etykiety_yolo


def zbuduj_zestaw_danych():
    print("Generowanie syntetycznego zbioru danych z logotypami oraz stosowanie preprocessingu...")
    przygotuj_katalogi()

    for idx in range(180):
        surowy, przetworzony, etykiety = generuj_probke()
        if not etykiety:
            continue
        podzbior = "train" if idx < 145 else "val"
        nazwa = f"logo_{idx:04d}"
        cv2.imwrite(os.path.join(KATALOG_BAZOWY, "images", podzbior, f"{nazwa}.jpg"), przetworzony)
        with open(os.path.join(KATALOG_BAZOWY, "labels", podzbior, f"{nazwa}.txt"), "w", encoding="utf-8") as plik:
            plik.write("\n".join(etykiety))

    for idx in range(16):
        surowy, przetworzony, _ = generuj_probke()
        nazwa = f"test_{idx:03d}.jpg"
        cv2.imwrite(os.path.join(KATALOG_BAZOWY, "images", "test_surowe", nazwa), surowy)
        cv2.imwrite(os.path.join(KATALOG_BAZOWY, "images", "test_przetworzone", nazwa), przetworzony)

    sciezka_yaml = os.path.join(KATALOG_BAZOWY, "dane_logo.yaml")
    with open(sciezka_yaml, "w", encoding="utf-8") as f:
        f.write(f"path: {os.path.abspath(KATALOG_BAZOWY)}\n")
        f.write("train: images/train\n")
        f.write("val: images/val\n")
        f.write(f"nc: {len(KLASY_LOGO)}\n")
        f.write(f"names: {KLASY_LOGO}\n")

    return sciezka_yaml


def wizualizuj_preprocessing():
    surowy = cv2.imread(os.path.join(KATALOG_BAZOWY, "images", "test_surowe", "test_000.jpg"))
    przetworzony = cv2.imread(os.path.join(KATALOG_BAZOWY, "images", "test_przetworzone", "test_000.jpg"))

    fig, osie = plt.subplots(1, 2, figsize=(12, 5))
    osie[0].imshow(cv2.cvtColor(surowy, cv2.COLOR_BGR2RGB))
    osie[0].set_title("Obraz surowy (niski kontrast, szum, rozmycie)")
    osie[0].axis("off")

    osie[1].imshow(cv2.cvtColor(przetworzony, cv2.COLOR_BGR2RGB))
    osie[1].set_title("Obraz po preprocessingu (Bilateral + CLAHE + Wyostrzanie)")
    osie[1].axis("off")

    plt.tight_layout()
    plt.show()


def trenuj_i_ocen_model(sciezka_yaml):
    urzadzenie = 0 if torch.cuda.is_available() else "cpu"
    print(f"Rozpoczynanie treningu modelu YOLOv8n na urzadzeniu: {urzadzenie}...")

    model = YOLO("yolov8n.pt")
    model.train(
        data=sciezka_yaml,
        epochs=20,
        imgsz=416,
        batch=16,
        device=urzadzenie,
        project="wyniki_detekcji",
        name="trening_logo",
        exist_ok=True,
        verbose=False,
    )

    metryki = model.val(data=sciezka_yaml, verbose=False)
    print(f"Wyniki walidacji -> mAP50: {metryki.box.map50:.4f} | mAP50-95: {metryki.box.map:.4f}")
    return model


def porownaj_detekcje_i_wizualizuj(model):
    katalog_surowe = os.path.join(KATALOG_BAZOWY, "images", "test_surowe")
    katalog_przetw = os.path.join(KATALOG_BAZOWY, "images", "test_przetworzone")

    wyniki_surowe = model.predict(source=katalog_surowe, conf=0.25, verbose=False)
    wyniki_przetw = model.predict(source=katalog_przetw, conf=0.25, verbose=False)

    pewnosc_surowe = [float(c) for r in wyniki_surowe for c in r.boxes.conf]
    pewnosc_przetw = [float(c) for r in wyniki_przetw for c in r.boxes.conf]

    srednia_sur = np.mean(pewnosc_surowe) if pewnosc_surowe else 0.0
    srednia_prz = np.mean(pewnosc_przetw) if pewnosc_przetw else 0.0

    print(f"Wykryte logotypy bez preprocessingu: {len(pewnosc_surowe)} (srednia pewnosc: {srednia_sur:.2%})")
    print(f"Wykryte logotypy po preprocessingu:  {len(pewnosc_przetw)} (srednia pewnosc: {srednia_prz:.2%})")

    fig, osie = plt.subplots(2, 4, figsize=(16, 8))
    for i in range(4):
        obraz_sur = wyniki_surowe[i].plot()
        obraz_prz = wyniki_przetw[i].plot()

        osie[0, i].imshow(cv2.cvtColor(obraz_sur, cv2.COLOR_BGR2RGB))
        osie[0, i].set_title(f"Bez preprocessingu #{i + 1}")
        osie[0, i].axis("off")

        osie[1, i].imshow(cv2.cvtColor(obraz_prz, cv2.COLOR_BGR2RGB))
        osie[1, i].set_title(f"Po preprocessingu #{i + 1}")
        osie[1, i].axis("off")

    plt.tight_layout()
    plt.show()

    wykres_treningu = os.path.join("wyniki_detekcji", "trening_logo", "results.png")
    if os.path.exists(wykres_treningu):
        img_wyniki = cv2.imread(wykres_treningu)
        plt.figure(figsize=(14, 7))
        plt.imshow(cv2.cvtColor(img_wyniki, cv2.COLOR_BGR2RGB))
        plt.title("Krzywe uczenia i metryki detekcji logotypow")
        plt.axis("off")
        plt.show()


if __name__ == "__main__":
    plik_konfiguracyjny = zbuduj_zestaw_danych()
    wizualizuj_preprocessing()
    wytrenowany_model = trenuj_i_ocen_model(plik_konfiguracyjny)
    porownaj_detekcje_i_wizualizuj(wytrenowany_model)
```
